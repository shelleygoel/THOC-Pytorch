```python
def create_dataset_from_dataframe(
    raw_df: pd.DataFrame,
    input_feature_columns: list[int],
    label_column: int,
    input_width: int,
    label_width: int,
    label_offset: int,
    batch_size: int = 32,
) -> tf.data.Dataset:
    total_width = input_width + label_offset + label_width - 1
    label_start = input_width + label_offset - 1
    label_column_index = [
        i for i, column in enumerate(df.columns) if column == label_column
    ]
    feature_column_index = [
        i for i, column in enumerate(df.columns) if column in input_feature_columns
    ]
    time_column_index = [i for i, column in enumerate(df.columns) if column == "time"]
    data = tf.keras.utils.timeseries_dataset_from_array(
        data=raw_df,
        targets=None,
        sequence_length=total_width,
        sequence_stride=1,
        batch_size=batch_size,
        shuffle=False,
    )

    def split_dataset(data) -> tuple[tf.data.Dataset, tf.data.Dataset]:
        input_column_index_tensor = tf.constant(feature_column_index, dtype=tf.int32)
        label_column_index_tensor = tf.constant(label_column_index, dtype=tf.int32)
        time_column_index_tensor = tf.constant(time_column_index, dtype=tf.int32)
        input = tf.gather(
            data[:, slice(0, input_width), :], input_column_index_tensor, axis=-1
        )
        label = tf.gather(
            data[:, slice(label_start, None), :], label_column_index_tensor, axis=-1
        )
        input_time = tf.gather(
            data[:, slice(0, input_width), :], time_column_index_tensor, axis=-1
        )
        label_time = tf.gather(
            data[:, slice(label_start, None), :], time_column_index_tensor, axis=-1
        )
        input.set_shape([None, input_width, len(feature_column_index)])
        label.set_shape([None, label_width, len(label_column_index)])
        input_time.set_shape([None, input_width, 1])
        label_time.set_shape([None, label_width, 1])

        # print(input_time.shape)
        # input_time.set_shape([None, input_width, 1])
        # label_time.set_shape([None, label_width, 1])
        return (input, input_time), (label, label_time)
        # return input, label

    return data.map(split_dataset)

df["time"] = df.index
input_feature_columns = [col for col in df.columns if col != "time"]
input_width = 100
label_column = 0
label_width = 1
label_offset = 1
train_dataset = create_dataset_from_dataframe(
    df, input_feature_columns, label_column, input_width, label_width, label_offset
)
train_size = int(0.8 * len(train_dataset))
val_data = train_dataset.skip(train_size)
train_data = train_dataset.take(train_size)
train_dataset.element_spec
```

```python
class MyModel(tf.keras.Model):

    def __init__(self, base_model: tf.keras.Model):
        super().__init__()
        self.base_model = base_model

    def call(self, features: tf.data.Dataset, training: bool = False):
        return self.base_model(features, training=training)
    
    def train_step(self, train_data: tf.data.Dataset):
        """
        Runs one step of training. It is called by the fit method of tf.keras.Model.


        """
        (features, _), (labels, _) = train_data
        # features, labels = train_data

        with tf.GradientTape() as tape:
            y_pred = self(features, training=True)
            loss = self.compiled_loss(labels, y_pred)
        
        trainable_vars = self.trainable_variables
        gradients = tape.gradient(loss, trainable_vars)
        self.optimizer.apply_gradients(zip(gradients, trainable_vars))
        # self.compiled_metrics.update_state(labels, y_pred)
        for metric in self.metrics:
            metric.update_state(labels, y_pred)

        return {m.name: m.result() for m in self.metrics}

multi_step_dense = tf.keras.Sequential(
    [
        # Shape: (time, features) => (time*features)
        tf.keras.layers.Flatten(input_shape=(input_width, 25)),
        tf.keras.layers.Dense(units=32, activation="relu"),
        tf.keras.layers.Dense(units=32, activation="relu"),
        tf.keras.layers.Dense(units=1),
        # Add back the time dimension.
        # Shape: (outputs) => (1, outputs)
        tf.keras.layers.Reshape([1, -1]),
        # tf.keras.layers.Lambda(lambda x: tf.expand_dims(x, axis=1))
    ]
)
my_model = MyModel(multi_step_dense)
patience = 2
MAX_EPOCHS = 20

early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss", patience=patience, mode="min"
)

my_model.compile(
    loss=tf.keras.losses.MeanSquaredError(),
    optimizer=tf.keras.optimizers.Adam(),
    metrics=[tf.keras.metrics.MeanAbsoluteError()],
)

history = my_model.fit(
    train_data, epochs=MAX_EPOCHS, )
# validation_data=val_data, callbacks=[early_stopping]
# )

```

```python
def get_batch_predictions(test_dataset: tf.data.Dataset) -> pd.DataFrame:
    test_pred_df_list = []

    for batch in test_dataset:
        (test_features, test_features_time), (test_labels, test_labels_time) = batch
        y_test_pred = my_model(test_features)
        df_test_pred = pd.DataFrame({
            'time': test_labels_time.numpy().reshape((-1)),
            'label': test_labels.numpy().reshape((-1)),
            'pred': y_test_pred.numpy().reshape((-1)),
        })
        test_pred_df_list.append(df_test_pred)

    test_pred_df = pd.concat(test_pred_df_list)
    return test_pred_df
```

```python

df_test["time"] = df_test.index
test_dataset = create_dataset_from_dataframe(
    df_test,input_feature_columns, label_column, input_width, label_width, label_offset, batch_size=1000
)
test_pred_df = get_batch_predictions(test_dataset)

```
