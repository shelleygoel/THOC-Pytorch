```python
def create_dataset_from_dataframe(
    raw_df: pd.DataFrame,
    input_feature_columns: list[int],
    label_column: int,
    input_width: int,
    label_width: int,
    label_offset: int,
    batch_size: int = 32,
) -> tf.data.Dataset:
    total_width = input_width + label_offset + label_width - 1
    label_start = input_width + label_offset - 1
    label_column_index = [
        i for i, column in enumerate(df.columns) if column == label_column
    ]
    feature_column_index = [
        i for i, column in enumerate(df.columns) if column in input_feature_columns
    ]
    time_column_index = [i for i, column in enumerate(df.columns) if column == "time"]
    data = tf.keras.utils.timeseries_dataset_from_array(
        data=raw_df,
        targets=None,
        sequence_length=total_width,
        sequence_stride=1,
        batch_size=batch_size,
        shuffle=False,
    )

    def split_dataset(data) -> tuple[tf.data.Dataset, tf.data.Dataset]:
        input_column_index_tensor = tf.constant(feature_column_index, dtype=tf.int32)
        label_column_index_tensor = tf.constant(label_column_index, dtype=tf.int32)
        time_column_index_tensor = tf.constant(time_column_index, dtype=tf.int32)
        input = tf.gather(
            data[:, slice(0, input_width), :], input_column_index_tensor, axis=-1
        )
        label = tf.gather(
            data[:, slice(label_start, None), :], label_column_index_tensor, axis=-1
        )
        input_time = tf.gather(
            data[:, slice(0, input_width), :], time_column_index_tensor, axis=-1
        )
        label_time = tf.gather(
            data[:, slice(label_start, None), :], time_column_index_tensor, axis=-1
        )
        input.set_shape([None, input_width, len(feature_column_index)])
        label.set_shape([None, label_width, len(label_column_index)])
        input_time.set_shape([None, input_width, 1])
        label_time.set_shape([None, label_width, 1])

        # print(input_time.shape)
        # input_time.set_shape([None, input_width, 1])
        # label_time.set_shape([None, label_width, 1])
        return (input, input_time), (label, label_time)
        # return input, label

    return data.map(split_dataset)

df["time"] = df.index
input_feature_columns = [col for col in df.columns if col != "time"]
input_width = 100
label_column = 0
label_width = 1
label_offset = 1
train_dataset = create_dataset_from_dataframe(
    df, input_feature_columns, label_column, input_width, label_width, label_offset
)
train_size = int(0.8 * len(train_dataset))
val_data = train_dataset.skip(train_size)
train_data = train_dataset.take(train_size)
train_dataset.element_spec
```

```python
class MyModel(tf.keras.Model):

    def __init__(self, base_model: tf.keras.Model):
        super().__init__()
        self.base_model = base_model

    def call(self, features: tf.data.Dataset, training: bool = False):
        return self.base_model(features, training=training)
    
    def train_step(self, train_data: tf.data.Dataset):
        """
        Runs one step of training. It is called by the fit method of tf.keras.Model.


        """
        (features, _), (labels, _) = train_data
        # features, labels = train_data

        with tf.GradientTape() as tape:
            y_pred = self(features, training=True)
            loss = self.compiled_loss(labels, y_pred)
        
        trainable_vars = self.trainable_variables
        gradients = tape.gradient(loss, trainable_vars)
        self.optimizer.apply_gradients(zip(gradients, trainable_vars))
        # self.compiled_metrics.update_state(labels, y_pred)
        for metric in self.metrics:
            metric.update_state(labels, y_pred)

        return {m.name: m.result() for m in self.metrics}

multi_step_dense = tf.keras.Sequential(
    [
        # Shape: (time, features) => (time*features)
        tf.keras.layers.Flatten(input_shape=(input_width, 25)),
        tf.keras.layers.Dense(units=32, activation="relu"),
        tf.keras.layers.Dense(units=32, activation="relu"),
        tf.keras.layers.Dense(units=1),
        # Add back the time dimension.
        # Shape: (outputs) => (1, outputs)
        tf.keras.layers.Reshape([1, -1]),
        # tf.keras.layers.Lambda(lambda x: tf.expand_dims(x, axis=1))
    ]
)
my_model = MyModel(multi_step_dense)
patience = 2
MAX_EPOCHS = 20

early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss", patience=patience, mode="min"
)

my_model.compile(
    loss=tf.keras.losses.MeanSquaredError(),
    optimizer=tf.keras.optimizers.Adam(),
    metrics=[tf.keras.metrics.MeanAbsoluteError()],
)

history = my_model.fit(
    train_data, epochs=MAX_EPOCHS, )
# validation_data=val_data, callbacks=[early_stopping]
# )

```

```python
def get_batch_predictions(test_dataset: tf.data.Dataset) -> pd.DataFrame:
    test_pred_df_list = []

    for batch in test_dataset:
        (test_features, test_features_time), (test_labels, test_labels_time) = batch
        y_test_pred = my_model(test_features)
        df_test_pred = pd.DataFrame({
            'time': test_labels_time.numpy().reshape((-1)),
            'label': test_labels.numpy().reshape((-1)),
            'pred': y_test_pred.numpy().reshape((-1)),
        })
        test_pred_df_list.append(df_test_pred)

    test_pred_df = pd.concat(test_pred_df_list)
    return test_pred_df
```

```python
from collections import namedtuple




def add_event_start_end(df: pd.DataFrame, timedelta: int) -> pd.DataFrame:
    df = df.copy()
    df = df.sort_values("time")
    df["time_diff"] = df["time"].diff(1).fillna(timedelta+1)
    df["event_start"] = df["time"].where(df["time_diff"] > timedelta, np.nan, inplace=False).ffill()

    df["time_lag_diff"] = df["time_diff"].shift(-1).fillna(timedelta+1)
    df["event_end"] = df["time"].where(df["time_lag_diff"] > timedelta, np.nan, inplace=False).bfill()

    return df

error_history_window = 2100
z = [2, 4, 6, 8, 10]
error_loss = [] #namedtuple("error_loss", ["z_value", "error_loss"])
error_sequences = np.array_split(test_pred_df[["time", "error"]].values, len(test_pred_df)// error_history_window)

all_anomalies = []
for seq in error_sequences:
    error_df = pd.DataFrame(seq, columns=["time", "error"])
    mean_error = error_df["error"].mean()
    std_error = error_df["error"].std() 
    # print(mean_error, std_error)
    for z_value in z:
        # print(z_value)
        error_threshold = mean_error + z_value * std_error
        nominal_errors = error_df[error_df["error"] < error_threshold]
        nominal_mean = nominal_errors["error"].mean()
        nominal_std = nominal_errors["error"].std()

        anomalous_errors = error_df[error_df["error"] >= error_threshold]
        anomalous_errors = add_event_start_end(anomalous_errors, timedelta=1)
        num_anomalous_points = anomalous_errors.shape[0]
        num_cont_anomalies = anomalous_errors["event_start"].nunique()
        greedy_penalty_term = max(1, num_anomalous_points + num_cont_anomalies**2)

        delta_mean = mean_error - nominal_mean
        delta_std = std_error - nominal_std
        if mean_error > 0 and std_error > 0:
            eloss = (delta_mean/mean_error + delta_std/std_error) / greedy_penalty_term 
        error_loss.append((z_value, error_threshold, eloss))
        # find continuous sequences of anomalous errors

    error_loss_df = pd.DataFrame(error_loss, columns=["z_value", "error_threshold", "error_loss"])
    seq_err_threshold = error_loss_df.loc[error_loss_df["error_loss"].argmax(), "error_threshold"]
    seq_anomalous_errors = error_df[error_df["error"] >= seq_err_threshold]
    seq_anomalous_errors = add_event_start_end(seq_anomalous_errors, timedelta=1)
    max_error = seq_anomalous_errors.groupby("event_start")["error"].max().rename("score")
    score = (max_error - seq_err_threshold) / (mean_error + std_error)
    seq_anomalous_errors = seq_anomalous_errors.merge(score, left_on="event_start", right_index=True)
    all_anomalies.append(seq_anomalous_errors)





        # error_loss(z_value, (mean_error + z_value * std_error))
```

```python
unique_events = final_test_pred_df[final_test_pred_df["is_error"]].drop_duplicates(subset=["event_start"])
event_tolerance = 10 # in minutes
is_anomaly_true_list = []
true_anomaly_event_list = []
for row in unique_events.iterrows():
    is_anomaly = False
    true_anomaly_event = []
    event_start = row[1]["event_start"]
    event_end = row[1]["event_end"]
    if event_end - event_start < event_tolerance:
        is_anomaly_true_list.append(is_anomaly)
        true_anomaly_event_list.append([()])
        continue
    for true_positive_seq in ch_anomalies["anomaly_sequences"].values:
        true_start = true_positive_seq[0]
        true_end = true_positive_seq[1]
        is_case1 = (event_start > true_start - event_tolerance) and (event_end > true_start) and (event_end < true_end + event_tolerance)
        is_case2 = (event_start > true_start) and (event_start < true_end) and (event_end < true_end + event_tolerance)
        if is_case1 or is_case2:
            is_anomaly = True
            true_anomaly_event.append((true_start, true_end))
            break
    is_anomaly_true_list.append(is_anomaly)
    true_anomaly_event_list.append(true_anomaly_event)


unique_events["is_anomaly_true"] = is_anomaly_true_list
unique_events["true_anomaly_event"] = true_anomaly_event_list
unique_events["true_anomaly_event"] = unique_events["true_anomaly_event"].apply(safe_eval)
unique_events = unique_events.explode("true_anomaly_event")
```
